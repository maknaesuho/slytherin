## Cleaning Final Web Data

```python
df_final_web_data.head()
```

```python
df_final_web_data['process_step'].value_counts()
```

Assigning number values to process_step and date_time:

```python
step_mapping = {
    'start': 0,
    'step_1': 1,
    'step_2': 2,
    'step_3': 3,
    'confirm': 4
}

df_final_web_data['process_step'] = df_final_web_data['process_step'].map(step_mapping)
```

```python
df_final_web_data['date_time'] = pd.to_datetime(df_final_web_data['date_time'])

# Extract year, month, day, and the full time (HH:MM:SS)
df_final_web_data['year'] = df_final_web_data['date_time'].dt.year
df_final_web_data['month'] = df_final_web_data['date_time'].dt.month
df_final_web_data['day'] = df_final_web_data['date_time'].dt.day
df_final_web_data['time'] = df_final_web_data['date_time'].dt.time
```

```python
# Build a client-level analysis dataset
# One row = one cleaned Test or Control participant

assert df_final_demo["client_id"].is_unique

# Keep only web events from cleaned experiment participants
web_participants = df_final_web_data.loc[
    df_final_web_data["client_id"].isin(
        df_final_demo["client_id"]
    )
].copy()

# Convert timestamps only in this analysis copy
web_participants["date_time"] = pd.to_datetime(
    web_participants["date_time"],
    errors="coerce"
)

# Count unique visits and total recorded events per client
visit_event_metrics = (
    web_participants.groupby("client_id")
    .agg(
        number_of_visits=("visit_id", "nunique"),
        number_of_events=("client_id", "size")
    )
    .reset_index()
)

# Count unique visits that reached the confirmation step
completed_visits = (
    web_participants.loc[web_participants["process_step"] == 4]
    .dropna(subset=["visit_id"])
    .groupby("client_id")["visit_id"]
    .nunique()
    .rename("completed_visits")
    .reset_index()
)


# Calculate duration separately inside every visit
web_participants["date_time"] = pd.to_datetime(web_participants["date_time"])

visit_durations = (
    web_participants
    .dropna(subset=["visit_id", "date_time"])
    .groupby(["client_id", "visit_id"])
    .agg(
        visit_start=("date_time", "min"),
        visit_end=("date_time", "max")
    )
    .reset_index()
)

visit_durations["visit_duration_seconds"] = (
    visit_durations["visit_end"]
    - visit_durations["visit_start"]
).dt.total_seconds()

# Calculate the average visit duration for every client
average_duration = (
    visit_durations.groupby("client_id")
    .agg(
        average_visit_duration_seconds=(
            "visit_duration_seconds",
            "mean"
        )
    )
    .reset_index()
)


# Create the final client-level DataFrame

df_v1 = (
    df_final_demo
    .merge(
        visit_event_metrics,
        on="client_id",
        how="left",
        validate="one_to_one"
    )

    .merge(
        completed_visits,
        on="client_id",
        how="left",
        validate="one_to_one"
    )

    .merge(
        average_duration,
        on="client_id",
        how="left",
        validate="one_to_one"
    )
)

# Clients without web events receive zero counts
count_columns = [
    "number_of_visits",
    "number_of_events",
    "completed_visits"
]

df_v1[count_columns] = (
    df_v1[count_columns]
    .fillna(0)
    .astype(int)
)

# Keep duration as NaN when no valid timed visit exists
assert df_v1["client_id"].is_unique
assert len(df_v1) == len(df_final_demo)

display(df_v1.head())

display(
    df_v1[
        [
            "number_of_visits",
            "number_of_events",
            "completed_visits",
            "average_visit_duration_seconds"
        ]
    ].describe()
)

df_v1.to_csv("df_v1.csv", index=False)
```
