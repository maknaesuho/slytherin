# Data Cleaning

Stack the rows on top of each other using pd.concat() and reset the index:

```python
df_final_web_data = pd.concat([d2_1, d2_2], ignore_index=True)
df_final_web_data 
```

Merging d1 and d3 due to same row number, using client_id as unique identifier. d3 is the list revealing which clients were part of the grand experiment so we need to remove the ones which were not:

```python
df_final_demo = df_final_demo.merge(df_experiment_clients, how='inner', on='client_id')

df_final_demo = df_final_demo.dropna(subset=['Variation'])
```
